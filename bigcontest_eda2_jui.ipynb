```python
import pandas as pd  
import numpy as np 
import re, json
import plotly.express as px
import plotly.graph_objects as go 
from plotly.subplots import make_subplots
```

```python
df = pd.read_csv("./data/seongdong_final.csv")

df["기준년월"] = pd.to_datetime(df["기준년월"].astype(str) + "01", format="%Y%m%d", errors="coerce")
df["ym"] = df["기준년월"].dt.to_period("M").astype(str)

df["개설일"] = pd.to_datetime(df["개설일"], format="%Y%m%d", errors="coerce")
df["폐업일"] = pd.to_datetime(df["폐업일"], format="%Y%m%d", errors="coerce")

df["폐업유무"] = df["폐업일"].notna().astype(int)
df["폐업라벨"] = df["폐업유무"].map({0: "영업", 1: "폐업"})

def month_diff(a, b):
    if pd.isna(a) or pd.isna(b): return np.nan
    return (a.year - b.year)*12 + (a.month - b.month)

df["운영개월수"] = [month_diff(a, b) for a, b in zip(df["기준년월"], df["개설일"])]
```

```python
STATUS_OPTIONS = [("전체", None), ("영업", 0), ("폐업", 1)]

def normalize_props(df, cols):
    out = df.copy()
    for c in cols:
        if c not in out.columns: 
            continue
        s = pd.to_numeric(out[c], errors="coerce")
        # sentinel 제거
        s = s.mask(s <= -1e5, np.nan).mask(s >= 1e6, np.nan)
        # 스케일 판정: 0~1인지 0~100인지
        q50 = s.quantile(0.5)
        q95 = s.quantile(0.95)
        # 케이스 A) 이미 0~1 스케일로 보임
        if pd.notna(q95) and q95 <= 1.2:
            pass
        # 케이스 B) 0~100 스케일로 보임
        elif pd.notna(q95) and (1 < q95 <= 1000):
            s = s / 100.0
        # 범위 밖 제거
        s = s.where((0 <= s) & (s <= 1))
        out[c] = s
    return out
```

```python
# 상태(전체/영업/폐업)별 월평균과 월별 표본수 반환.
# roll: 이동평균 윈도우(0이면 생략) / min_n: 월별 유효 표본이 이보다 적으면 NaN 마스킹
def build_series_by_status(df, cols, status_flag, roll=3, min_n=None):
    if status_flag is None:
        sub = df.dropna(subset=["ym"] + cols).copy()
    else:
        sub = df[df["폐업유무"] == status_flag].dropna(subset=["ym"] + cols).copy()

    # 월별 평균 & 표본수
    mean_df = sub.groupby("ym")[cols].mean()
    n_s = sub.groupby("ym")[cols[0]].size()  # 임의 한 컬럼 기준으로 n

    # 표본 부족한 달은 NaN
    mean_df = mean_df.where(n_s >= min_n)

    # 이동평균
    if roll and roll > 1:
        mean_df = mean_df.rolling(roll, min_periods=1).mean()

    idx = pd.to_datetime(mean_df.index.astype(str) + "-01", errors="coerce")
    mean_df.index = idx
    return mean_df, n_s.reindex(mean_df.index, fill_value=0)
```

```python
# 여러 지표를 한 그래프에 라인 차트 - 드롭다운
def plot_group_lines_with_status(df, cols, title, roll=3, min_n=None, ytitle="값(평균)"):
    fig = go.Figure()
    vis_sets = []
    traces_per_state = len(cols)
    total_traces = traces_per_state * len(STATUS_OPTIONS)

    for label, flag in STATUS_OPTIONS:
        mean_df, _ = build_series_by_status(df, cols, flag, roll=roll, min_n=min_n)
        for c in cols:
            fig.add_trace(go.Scatter(x=mean_df.index, y=mean_df[c].values,
                                     mode="lines+markers", name=f"{c} — {label}"))

    # visible 배열
    for si, (label, _) in enumerate(STATUS_OPTIONS):
        vis = [False] * total_traces
        start = si * traces_per_state
        for i in range(traces_per_state):
            vis[start + i] = True
        vis_sets.append((label, vis))

    buttons = [dict(label=label, method="update",
                    args=[{"visible": vis},{"title": f"{title} ({label})"}])for label, vis in vis_sets]

    fig.update_layout(title=f"{title} (전체)", xaxis_title="월", yaxis_title=ytitle,
                      xaxis=dict(rangeslider=dict(visible=True), dtick="M1", tickformat="%y.%m"),
                      legend=dict(orientation="v", x=1.02, y=1), margin=dict(l=40, r=200, t=60, b=40), 
                      updatemenus=[dict(type="dropdown", x=1.0, xanchor="right", y=1.15, buttons=buttons)])
    
    return fig
```

```python
# 단일 지표에 대해 Bar(1-평균=안정도) + Line(>risk_thr 비중) — 드롭다운
def plot_group_barline_risk(df, col, risk_thr=0.10, roll=3, min_n=None):
    fig = make_subplots(specs=[[{"secondary_y": True}]])
    traces = []

    for label, flag in STATUS_OPTIONS:
        mean_df, n_s = build_series_by_status(df, [col], flag, roll=roll, min_n=min_n)
        s_mean = mean_df[col]
        # 위험비중(> thr)
        if flag is None:
            sub = df.dropna(subset=["ym", col]).copy()
        else:
            sub = df[df["폐업유무"] == flag].dropna(subset=["ym", col]).copy()
        risk_s = sub.groupby("ym").apply(lambda x: (x[col] > risk_thr).mean())
        # 동일 인덱스로 맞추기 + 이동평균
        risk_s.index = pd.to_datetime(risk_s.index.astype(str) + "-01")
        risk_s = risk_s.reindex(s_mean.index)
        if roll and roll > 1:
            risk_s = risk_s.rolling(roll, min_periods=1).mean()

        # Bar: 안정도(1-평균)
        traces.append(("bar", go.Bar(x=s_mean.index, y=(1 - s_mean.values),
                                     name=f"{label} — 안정도", opacity=0.35)))
        # Line: 위험비중(%)
        traces.append(("line", go.Scatter(x=risk_s.index, y=(risk_s.values * 100),
                                          mode="lines+markers", name=f"{label} — 위험비중")))

    # 추가 & 드롭다운
    for i, (kind, tr) in enumerate(traces):
        fig.add_trace(tr, secondary_y=(kind == "line"))

    total = len(traces)
    per_state = 2
    buttons = []
    for si, (label, _) in enumerate(STATUS_OPTIONS):
        vis = [False]*total
        vis[si*per_state] = True
        vis[si*per_state + 1] = True
        buttons.append(dict(label=label, method="update",
                            args=[{"visible": vis},{"title": f"[시계열] {col} : 안정도(Bar) + 위험비중(Line) ({label})",
                                                    "yaxis": {"title": "안정도(높을수록 양호)", "range": [0,1]},
                                                    "yaxis2": {"title": "위험비중", "ticksuffix": "%"}}]))

    fig.update_layout(title=f"[시계열] {col} : 안정도(Bar) + 위험비중(Line) (전체)", xaxis_title="월",
                      xaxis=dict(rangeslider=dict(visible=True), dtick="M1", tickformat="%y.%m"),
                      legend=dict(orientation="v", x=1.02, y=1), margin=dict(l=40, r=200, t=60, b=40),
                      updatemenus=[dict(type="dropdown", x=1.0, xanchor="right", y=1.15, buttons=buttons)])
    fig.update_yaxes(secondary_y=False, title_text="안정도(높을수록 양호)", range=[0,1])
    fig.update_yaxes(secondary_y=True,  title_text="위험비중", ticksuffix="%")
    
    return fig
```

```python
# 여러 구성비(합≈1인 컬럼들)를 100% 스택 면적으로 (상태 드롭다운)
def plot_stacked_percent_with_status(df, cols, title, roll=3, min_n=None):
    fig = go.Figure()
    vis_sets = []
    traces_per_state = len(cols)
    total_traces = traces_per_state * len(STATUS_OPTIONS)

    for label, flag in STATUS_OPTIONS:
        mean_df, _ = build_series_by_status(df, cols, flag, roll=roll, min_n=min_n)
        # 행(row) 합으로 100% 정규화
        row_sum = mean_df.sum(axis=1)
        pct_df = mean_df.div(row_sum, axis=0) * 100.0
        for c in cols:
            fig.add_trace(go.Scatter(x=pct_df.index, y=pct_df[c].values, mode="lines",
                                     stackgroup=f"{label}_pct", name=f"{c} — {label}"))

    for si, (label, _) in enumerate(STATUS_OPTIONS):
        vis = [False] * total_traces
        start = si * traces_per_state
        for i in range(traces_per_state):
            vis[start + i] = True
        vis_sets.append((label, vis))

    buttons = [dict(label=label, method="update",
                    args=[{"visible": vis},{"title": f"{title} ({label})"}])for label, vis in vis_sets]

    fig.update_layout(title=f"{title} (전체)", xaxis_title="월", yaxis_title="구성비(%)",
                      xaxis=dict(rangeslider=dict(visible=True), dtick="M1", tickformat="%y.%m"),
                      yaxis=dict(range=[0,100]),
                      legend=dict(orientation="v", x=1.02, y=1), margin=dict(l=40, r=200, t=60, b=40),
                      updatemenus=[dict(type="dropdown", x=1.0, xanchor="right", y=1.15, buttons=buttons)])
    
    return fig
```

```python
# 운영개월수 기반 집계 함수
def summarize_by_tenure(df, col, min_n=None, roll=3, max_month=120):
    d = df.copy()

    d = d[pd.notna(d["운영개월수"])].copy()
    d["oper_m"] = d["운영개월수"].astype(int)
    d = d[(d["oper_m"] >= 0) & (d["oper_m"] <= max_month)]

    # 월별-상태별 평균, 표본
    gp = d.groupby(["oper_m","폐업유무"])[col].agg(["mean","size"]).reset_index()
    mean_pv = gp.pivot(index="oper_m", columns="폐업유무", values="mean").rename(columns={0:"영업",1:"폐업"})
    n_pv = gp.pivot(index="oper_m", columns="폐업유무", values="size").rename(columns={0:"영업",1:"폐업"}).fillna(0)
    n_total = n_pv.sum(axis=1).rename("N")

    # 표본 하한 적용
    mean_pv = mean_pv.where(n_total >= min_n)

    # 이동평균(운영개월수 기준)
    if roll and roll > 1:
        mean_pv = mean_pv.rolling(roll, min_periods=1).mean()
        n_total = n_total.rolling(roll, min_periods=1).mean()

    mean_pv.index.name = "운영개월수"
    out = mean_pv.copy()
    out["N"] = n_total
    return out

# 단일 지표 시각화 (영업/폐업 라인 + 표본수 보조축 바)
def plot_tenure_lines(df, col, min_n=None, roll=3, max_month=120, percent_axis=True):
    agg = summarize_by_tenure(df, col, min_n=min_n, roll=roll, max_month=max_month)

    fig = make_subplots(specs=[[{"secondary_y": True}]])
    # 라인: 영업/폐업 평균
    if "영업" in agg.columns:
        fig.add_trace(go.Scatter(x=agg.index, y=(agg["영업"]*100 if percent_axis else agg["영업"]),
                                 mode="lines+markers", name="영업"), secondary_y=False)
    if "폐업" in agg.columns:
        fig.add_trace(go.Scatter(x=agg.index, y=(agg["폐업"]*100 if percent_axis else agg["폐업"]),
                                 mode="lines+markers", name="폐업"), secondary_y=False)
    # 보조축: 전체 표본 수
    fig.add_trace(go.Bar(x=agg.index, y=agg["N"], name="표본수(N)", opacity=0.25),
                  secondary_y=True)

    fig.update_layout(title=f"[운영개월수] {col} — 영업 vs 폐업", xaxis_title="운영개월수(개월)",
                      legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
                      margin=dict(l=40, r=60, t=60, b=40), bargap=0.0)
    fig.update_yaxes(title_text=("값(%)" if percent_axis else "값"), secondary_y=False)
    fig.update_yaxes(title_text="표본수(N)", secondary_y=True, rangemode="tozero")
    return fig
```

```python
# 피벗 + 히트맵 
MIN_N = 10    
def make_admin_industry_matrix(df, metric, min_n=MIN_N):
    agg = (df.groupby(["업종","행정동"])[metric].agg(mean="mean", n="size").reset_index())
    pv_mean = agg.pivot(index="업종", columns="행정동", values="mean")
    pv_n = agg.pivot(index="업종", columns="행정동", values="n").fillna(0)
    # 표본 하한
    pv_mean = pv_mean.where(pv_n >= min_n)

    # 정렬: 표본 많은 순
    row_order = pv_n.sum(axis=1).sort_values(ascending=False).index
    col_order = pv_n.sum(axis=0).sort_values(ascending=False).index
    pv_mean = pv_mean.reindex(index=row_order, columns=col_order)
    pv_n    = pv_n.reindex(index=row_order, columns=col_order)
    return pv_mean, pv_n

def plot_small_heatmap(df, metric, title_suffix="", min_n=MIN_N, width=900, height=600):
    pv_mean, pv_n = make_admin_industry_matrix(df, metric, min_n=min_n)
    if pv_mean.notna().sum().sum() == 0:
        print(f"[WARN] '{metric}' 값이 모두 NaN 또는 표본(n<{min_n}) 부족. "
              f"우선 카운트 히트맵으로 폴백합니다.")
        z = pv_n.values
        colorscale = "Greys"
        colorbar_title = "표본수(n)"
        hovertemplate = "<b>%{y}</b> · <b>%{x}</b><br>n: %{z}<extra></extra>"
    else:
        z = pv_mean.values
        colorscale = "Reds"
        colorbar_title = "비중(0~1)"
        hovertemplate = "<b>%{y}</b> · <b>%{x}</b><br>비중: %{z:.2f}<extra></extra>"

    fig = go.Figure(data=go.Heatmap(z=z, x=pv_mean.columns.tolist(), y=pv_mean.index.tolist(), zmin=0, 
                                    zmax=(1 if colorscale=="Reds" else None), colorscale=colorscale, 
                                    colorbar=dict(title=colorbar_title), hovertemplate=hovertemplate))
    
    fig.update_layout(title=f"[업종 × 행정동] {metric}" + (f" — {title_suffix}" if title_suffix else ""),
                      xaxis_title="행정동", yaxis_title="업종",
                      xaxis=dict(tickangle=45, automargin=True), yaxis=dict(automargin=True), width=width, 
                      height=height, margin=dict(l=100, r=20, t=60, b=100), font=dict(size=11))
    return fig
```

```python
# 행/열(업종/행정동) 순서 고정: 전체 표본 수 많은 순 
cnt_all = df.groupby(["업종","행정동"]).size().reset_index(name="n")
row_order = cnt_all.groupby("업종")["n"].sum().sort_values(ascending=False).index.tolist()     # 업종 (y)
col_order = cnt_all.groupby("행정동")["n"].sum().sort_values(ascending=False).index.tolist()   # 행정동 (x)

# 상태별 피벗 생성 
MIN_N = 10  
def admin_industry_pivot(df_src, metric, flag=None, min_n=MIN_N):
    sub = df_src if flag is None else df_src[df_src["폐업유무"] == flag]
    agg = (sub.groupby(["업종","행정동"])[metric].agg(mean="mean", n="size").reset_index())
    pv_mean = (agg.pivot(index="업종", columns="행정동", values="mean").reindex(index=row_order, columns=col_order))
    pv_n = (agg.pivot(index="업종", columns="행정동", values="n").reindex(index=row_order, columns=col_order).fillna(0))
    pv_mean = pv_mean.where(pv_n >= min_n)   # 표본 하한
    return pv_mean, pv_n

# 영업/폐업 드롭다운 히트맵 (y=업종, x=행정동)
def plot_heatmap_with_status(df_src, metric, min_n=MIN_N, width=900, height=600):
    traces = []
    texts = []
    for label, flag in STATUS_OPTIONS:
        pv_mean, pv_n = admin_industry_pivot(df_src, metric, flag, min_n=min_n)
        z = pv_mean.to_numpy(dtype=float)

        # hover text 구성
        text = []
        for i in range(len(pv_mean.index)):
            row_txt = []
            for j in range(len(pv_mean.columns)):
                val = pv_mean.iat[i, j]
                n   = pv_n.iat[i, j] if not pd.isna(pv_n.iat[i, j]) else 0
                if pd.notna(val):
                    row_txt.append(f"{val*100:.2f}% (n={int(n)})")
                else:
                    row_txt.append("")
            text.append(row_txt)

        traces.append(z)
        texts.append(text)

    fig = go.Figure()
    for i, (label, _) in enumerate(STATUS_OPTIONS):
        fig.add_trace(go.Heatmap(z=traces[i], x=col_order, y=row_order, zmin=0, zmax=1, colorscale="Reds",
                                 text=texts[i], hovertemplate="%{y} · %{x}<br>%{text}<extra></extra>",
                                 colorbar=dict(title="비중", tickformat=".0%"), visible=(i == 0)))

    buttons = []
    for i, (label, _) in enumerate(STATUS_OPTIONS):
        vis = [j == i for j in range(len(STATUS_OPTIONS))]
        buttons.append(dict(label=f"상태: {label}", method="update",
                            args=[{"visible": vis}, {"title": f"[업종 × 행정동] {metric} — {label} (n≥{min_n})"}]))

    fig.update_layout(title=f"[업종 × 행정동] {metric} — 전체 (n≥{min_n})", xaxis_title="행정동", yaxis_title="업종",
                      xaxis=dict(tickangle=45, automargin=True), yaxis=dict(automargin=True), width=width, height=height,
                      margin=dict(l=110, r=20, t=60, b=100), font=dict(size=11),
                      updatemenus=[dict(type="dropdown", direction="down", buttons=buttons, 
                                        x=0.01, y=1.12, xanchor="left", yanchor="top", 
                                        bgcolor="white", bordercolor="#ccc", pad=dict(l=6,r=6,t=6,b=6))])
    return fig
```

```python
def same_month(a_dt: pd.Series, b_dt: pd.Series):
    return (a_dt.dt.year == b_dt.dt.year) & (a_dt.dt.month == b_dt.dt.month)

# 최근 윈도우
df["ym_dt"] = pd.to_datetime(df["ym"].astype(str) + "-01", errors="coerce")
cut_dt = df["ym_dt"].max() - pd.DateOffset(months=24)
recent = df[df["ym_dt"] >= cut_dt].copy()

recent["운영개월수"] = pd.to_numeric(recent["운영개월수"], errors="coerce")
recent = recent[(recent["운영개월수"].notna()) & (recent["운영개월수"] >= 0) & (recent["운영개월수"] <= 150)].copy()

# 코호트(최초 개설 월) → 가맹점번호 기준으로 안전 매핑
cohort_map = (df.groupby("가맹점번호")["개설일"].min().dt.to_period("M").astype(str))
recent["cohort"] = recent["가맹점번호"].map(cohort_map)

# 그 달 폐업(분자)
recent["close_event"] = same_month(recent["폐업일"], recent["ym_dt"]).astype(int)

# 분모: 그 달 활성(개설 ≤ t ≤ 폐업 or 폐업 NaT)
recent["active"] = ((recent["개설일"] <= recent["ym_dt"]) &
                    (recent["폐업일"].isna() | (recent["ym_dt"] <= recent["폐업일"]))).astype(int)

# 코호트×운영개월수별 폐업률 = (그 달 폐업 수) / (그 달 활성 노출 수)
g = (recent.dropna(subset=["cohort", "운영개월수"]).groupby(["cohort", "운영개월수"], as_index=False)
                                                  .agg(pos=("close_event", "sum"), n=("active", "sum")))
g = g[g["n"] > 0].copy()
g["rate"] = g["pos"] / g["n"]

fig = px.line(g.sort_values(["cohort", "운영개월수"]), x="운영개월수", y="rate", color="cohort", title="코호트별 폐업률 추이")
fig.update_yaxes(title="평균 폐업률", tickformat=".2%")
fig.show()
```

```python
# 분모/분자
df["active"] = ((df["개설일"] <= df["ym_dt"]) & (df["폐업일"].isna() | (df["ym_dt"] <= df["폐업일"]))).astype(int)
df["close_event"] = same_month(df["폐업일"], df["ym_dt"]).astype(int)

bins = [0,6,12,18,24,30,36,42,48,54,60,66,72,78,84,90,96,102,108]
labels = [f"{bins[i]}–{bins[i+1]-1}" for i in range(len(bins)-1)]
df["tenure_bin"] = pd.cut(df["운영개월수"], bins=bins, right=False, labels=labels)

# 구간별 폐업률 = (그 달 폐업 수) / (그 달 활성 노출 수)
g = (df.dropna(subset=["tenure_bin"]).groupby("tenure_bin", observed=True)
       .agg(pos=("close_event","sum"), n=("active","sum")).reset_index())
g["rate"] = g["pos"] / g["n"]
g = g[g["n"]>0].copy()

fig = px.bar(g, x="tenure_bin", y="rate", text=g["n"].map(lambda x: f"n={x:,}"),
             title="운영개월수 구간별 폐업률(막대=rate, 라벨=n)")
fig.update_traces(textposition="outside")
fig.update_yaxes(tickformat=".2%", title="폐업률")
fig.update_xaxes(title="운영개월수 구간(개월)")
fig.show()
```

```python
ym_p = df["기준년월"].dt.to_period("M")
open_p = df["개설일"].dt.to_period("M")
close_p = df["폐업일"].dt.to_period("M")

active_mask = (open_p <= ym_p) & (close_p.isna() | (ym_p <= close_p))
closed_mask = close_p.eq(ym_p)

base = df[["업종", "행정동"]].copy()
base["ym"] = ym_p.astype(str)     
base["active"] = active_mask.astype(int)
base["closed"] = closed_mask.astype(int)

grp = (base.groupby(["업종","행정동","ym"], as_index=False).agg(N_active=("active","sum"), C_close=("closed","sum")))

# 월별 폐업률 (라플라스 스무딩; α=β=1 권장)
alpha, beta = 1.0, 1.0
grp["close_rate"] = (grp["C_close"] + alpha) / (grp["N_active"] + alpha + beta)

# 전체 기간 평균 폐업률, 가중치 = 해당 달 활성 점포 수
agg = (grp.groupby(["업종","행정동"], as_index=False)
       .apply(lambda g: pd.Series({"avg_close_rate": np.average(g["close_rate"], weights=g["N_active"]) if g["N_active"].sum()>0 else np.nan,
                                   "months_covered": g.shape[0], "N_active_total": g["N_active"].sum()})).reset_index(drop=True))

# 희소 셀 마스킹 (표본이 너무 적은 조합은 NaN 처리) 
RATE_CMAP = px.colors.sequential.dense     
N_CMAP    = px.colors.sequential.Sunsetdark
MIN_MONTHS = 6          # 최소 월 수
MIN_ACTIVE_TOTAL = 10   # 전체 기간 활성 점포 수 합계 하한
mask = (agg["months_covered"] >= MIN_MONTHS) & (agg["N_active_total"] >= MIN_ACTIVE_TOTAL)
agg["avg_close_rate_masked"] = agg["avg_close_rate"].where(mask) 

# 히트맵: 평균 폐업률 
mat_rate = agg.pivot(index="업종", columns="행정동", values="avg_close_rate_masked")
row_order = mat_rate.mean(axis=1, skipna=True).sort_values(ascending=False).index
col_order = mat_rate.mean(axis=0, skipna=True).sort_values(ascending=False).index

fig1 = px.imshow(mat_rate, aspect="auto", color_continuous_scale=RATE_CMAP,
                 title=f"업종 × 행정동 평균 월별 폐업률 (n_months≥{MIN_MONTHS}, N_active_total≥{MIN_ACTIVE_TOTAL})")
fig1.update_coloraxes(colorbar_title="평균 폐업률")
fig1.update_xaxes(categoryorder="array", categoryarray=list(col_order))
fig1.update_yaxes(categoryorder="array", categoryarray=list(row_order))
fig1.update_layout(width=900, height=600)
fig1.show()

# 히트맵: 표본 규모(활성 점포 수 합계) 
mat_n = agg.pivot(index="업종", columns="행정동", values="N_active_total")

fig2 = px.imshow(mat_n, aspect="auto", color_continuous_scale=N_CMAP,
                 title="업종 × 행정동 표본 규모 (전체 기간 활성 점포 수 합계)")
fig2.update_coloraxes(colorbar_title="활성 점포 수 합계")
fig2.update_xaxes(categoryorder="array", categoryarray=list(col_order))
fig2.update_yaxes(categoryorder="array", categoryarray=list(row_order))
fig2.update_layout(width=1000, height=600)
fig2.show()
```

```python
GEOJSON_PATH   = "./data/administrative_district/4_서울시_행정동/서울_행정동_경계_2017.geojson"
FEATURE_ID_KEY = "properties.adm_nm"    
DONG_PREFIX    = "서울특별시 성동구 "  
DONG_COL       = "행정동"
LAT_COL        = "위도"
LON_COL        = "경도"
IND_COL        = "업종"
CLOSED_COL     = "폐업유무"           
TOP_INDS       =  None                
OTHER_LABEL    = "기타"

# 기본 전처리
df[DONG_COL] = DONG_PREFIX + df["행정동"].astype(str)

tmp = df.copy()

tmp[CLOSED_COL] = tmp["폐업일"].notna().astype(int)
tmp[LAT_COL] = pd.to_numeric(tmp[LAT_COL], errors="coerce")
tmp[LON_COL] = pd.to_numeric(tmp[LON_COL], errors="coerce")
tmp = tmp.dropna(subset=[DONG_COL, LAT_COL, LON_COL, IND_COL]).copy()

# 업종 상위 N개만 유지
if TOP_INDS is not None:
    keep_inds = tmp[IND_COL].value_counts().head(TOP_INDS).index.tolist()
    tmp[IND_COL] = np.where(tmp[IND_COL].isin(keep_inds), tmp[IND_COL], OTHER_LABEL)

# 행정동별 전체 가게 수
dong_count_all = tmp.groupby(DONG_COL).size().reset_index(name="store_cnt")

# 지도 중심
center_lat = float(tmp[LAT_COL].mean())
center_lon = float(tmp[LON_COL].mean())

# GeoJSON 로드
with open(GEOJSON_PATH, "r", encoding="utf-8") as f:
    gj = json.load(f)

# 업종 팔레트(겹치지 않도록 충분한 길이)
PALETTE = (px.colors.qualitative.Alphabet + px.colors.qualitative.Dark24 + px.colors.qualitative.Light24)

# 업종 목록
inds = tmp[IND_COL].value_counts().index.tolist()
if len(PALETTE) < len(inds):
    pass
color_map = {ind: PALETTE[i % len(PALETTE)] for i, ind in enumerate(inds)}

# Choropleth: 행정동별 가게 수
fig = px.choropleth_mapbox(dong_count_all, geojson=gj, locations=DONG_COL, featureidkey=FEATURE_ID_KEY,
                           color="store_cnt", color_continuous_scale="Blues", mapbox_style="carto-positron", 
                           zoom=13, center={"lat": center_lat, "lon": center_lon}, opacity=0.55, 
                           hover_data={DONG_COL: True, "store_cnt": True})
fig.update_traces(marker_line_width=1.2, marker_line_color="rgba(0,70,130,0.85)")

# 점 레이어(업종별 + 폐업 마커)
open_df = tmp[tmp[CLOSED_COL] == 0].copy()
close_df = tmp[tmp[CLOSED_COL] == 1].copy()

for ind in inds:
    col = color_map[ind]
    lg  = f"ind_{ind}"

    sub_open = open_df[open_df[IND_COL] == ind]
    if len(sub_open):
        fig.add_trace(go.Scattermapbox(lat=sub_open[LAT_COL], lon=sub_open[LON_COL], mode="markers", 
                                       marker=dict(size=8, color=col, opacity=0.85),
                                       name=str(ind), legendgroup=lg, showlegend=True,
                                       hovertemplate=f"<b>{ind}</b><extra>영업</extra>"))

    # 폐업 점(검은 외곽 + 업종색 내부)
    sub_close = close_df[close_df[IND_COL] == ind]
    if len(sub_close):
        fig.add_trace(go.Scattermapbox(lat=sub_close[LAT_COL], lon=sub_close[LON_COL], mode="markers",
                                       marker=dict(size=11, color="black", opacity=1.0),
                                       name=str(ind), legendgroup=lg, showlegend=False, hoverinfo="skip"))
        fig.add_trace(go.Scattermapbox(lat=sub_close[LAT_COL], lon=sub_close[LON_COL], mode="markers",
                                       marker=dict(size=8, color=col, opacity=0.95),
                                       name=str(ind), legendgroup=lg, showlegend=False,
                                       hovertemplate=f"<b>{ind}</b><extra>폐업</extra>"))

# Choropleth 드롭다운(전체/업종별로 채색값 전환)
choro_idx = None
for i, tr in enumerate(fig.data):
    if tr.type == "choroplethmapbox":
        choro_idx = i
        break
locations_order = list(fig.data[choro_idx].locations)

# 행정동별 카운트 벡터 준비 함수
def _counts_by_dong_for(_ind=None):
    if _ind is None:
        s = tmp.groupby(DONG_COL).size()
    else:
        s = tmp.loc[tmp[IND_COL] == _ind].groupby(DONG_COL).size()
    s = s.reindex(locations_order).fillna(0).astype(int)
    return s.values

z_all = _counts_by_dong_for(None)
z_by_ind = {ind: _counts_by_dong_for(ind) for ind in inds}

# 기본값: 전체
fig.data[choro_idx].z = z_all
fig.data[choro_idx].hovertemplate = "<b>%{location}</b><br>가게 수: %{z}<extra></extra>"

# 드롭다운 버튼
dropdown_buttons = [dict(label="전체", method="restyle", args=[{"z": [z_all]}, [choro_idx]])]
for ind in inds:
    dropdown_buttons.append(dict(label=str(ind), method="restyle", args=[{"z": [z_by_ind[ind]]}, [choro_idx]]))

fig.update_layout(title="행정동 가게 수 Choropleth + 업종별 점", width=1200, height=800,
                  margin=dict(l=10, r=220, t=60, b=80),
                  coloraxis_colorbar=dict(title="가게 수", orientation="h", x=0.50, y=0.02, xanchor="center", 
                                          yanchor="bottom", thickness=12, len=0.60, titleside="top"),
                  legend=dict(title="업종", orientation="v", x=1.02, y=1.0, xanchor="left", yanchor="top", 
                              bgcolor="rgba(255,255,255,0.85)", groupclick="togglegroup", itemdoubleclick="toggleothers", 
                              font=dict(size=11), itemsizing="constant", traceorder="normal"),
                  updatemenus=[dict(type="dropdown", direction="down", buttons=dropdown_buttons, 
                                    x=0.02, y=0.98, xanchor="left", yanchor="top", 
                                    pad=dict(l=6, r=6, t=6, b=6), bgcolor="white", bordercolor="#ccc")])
fig.show()
```
